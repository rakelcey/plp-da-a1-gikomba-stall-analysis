```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
highest_sale = max(daily_sales)
lowest_sale = min(daily_sales)

best_day_idx = daily_sales.index(highest_sale)
worst_day_idx = daily_sales.index(lowest_sale)

best_day_name = days[best_day_idx]
best_week_num = (best_day_idx // 7) + 1

worst_day_name = days[worst_day_idx]
worst_week_num = (worst_day_idx // 7) + 1

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Highest Sale: KES {highest_sale:,} ({best_day_name}, Week {best_week_num})")
print(f"Lowest Sale: KES {lowest_sale:,} ({worst_day_name}, Week {worst_week_num})")
```

```python
def week_totals(sales):
    """Calculates and returns the total sales for Week 1 and Week 2 using list slicing."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2

w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total: KES {w1_total:,}")
print(f"Week 2 Total: KES {w2_total:,}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

```python
def classify_day(amount, average):
    """Classifies a daily sales amount relative to the overall average."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    category = classify_day(amount, avg_sales)
    category_counts[category] += 1

print("Day Classifications Count:")
for cat, count in category_counts.items():
    print(f"- {cat}: {count} days")
```

```python
weekday_sales = [daily_sales[i] for i in range(len(daily_sales)) if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]]
weekend_sales = [daily_sales[i] for i in range(len(daily_sales)) if days[i] in ["Sat", "Sun"]]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)

diff_percentage = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend vs Weekday Difference: {diff_percentage:.2f}% higher on weekends")
```

```python
def get_reorder_items(stock_dict):
    """Returns a list of product names whose stock quantity is at or below the reorder level."""
    reorder_list = []
    for product, info in stock_dict.items():
        if info["qty"] <= info["reorder"]:
            reorder_list.append(product)
    return reorder_list

items_to_restock = get_reorder_items(stock)

print("Products to Reorder:")
for item in items_to_restock:
    print(f"- {item} (Current Qty: {stock[item]['qty']} kg, Reorder Level: {stock[item]['reorder']} kg)")

print("\nGross Margins by Product:")
for product, info in stock.items():
    margin = (info["price"] - info["cost"]) / info["price"] * 100
    print(f"- {product}: {margin:.1f}% gross margin")
```

```python
report = f"""
=========================================
      STALL PERFORMANCE REPORT
=========================================
Total 14-Day Sales     : KES {total_sales:,}
Average Daily Sales    : KES {avg_sales:,.2f}
Best Day Performance   : KES {highest_sale:,} ({best_day_name}, Week {best_week_num})
Worst Day Performance  : KES {lowest_sale:,} ({worst_day_name}, Week {worst_week_num})
-----------------------------------------
Week 1 Total Revenue   : KES {w1_total:,}
Week 2 Total Revenue   : KES {w2_total:,}
Week-on-Week Growth    : {wow_growth:.2f}%
-----------------------------------------
Average Weekday Sales  : KES {avg_weekday:,.2f}
Average Weekend Sales  : KES {avg_weekend:,.2f}
-----------------------------------------
Items Needing Restock  : {', '.join(items_to_restock)}
=========================================
"""
print(report)
```
